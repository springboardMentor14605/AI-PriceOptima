**CLEANING DATASET**

```python
import pandas as pd
df = pd.read_csv("retail_store_inventory_dataset.csv")
print(df.shape)
print(df.head())
```

```python
df.dtypes
```

```python
df["Date"] = pd.to_datetime(df["Date"], format="%d-%m-%Y")
```

```python
df.isnull().sum()
```

```python
df["Price"].fillna(df["Price"].median(), inplace=True)
```

```python
df["Weather_Condition"].fillna(df["Weather_Condition"].mode()[0], inplace=True)
```

```python
df.duplicated().sum()
df.drop_duplicates(inplace=True)
```

```python
df.columns = df.columns.str.lower().str.replace(" ", "_")
```

```python
df[df["inventory_level"] < 0]
df = df[df["inventory_level"] >= 0]
```

```python
df[df["units_sold"] > df["inventory_level"]]
```

```python
df[df["price"] <= 0]
```

```python
df[df["cost_price"] > df["price"]]
```

```python
df[(df["discount"] < 0) | (df["discount"] > 1)]
df["discount"] = df["discount"].clip(0,1)
```

```python
df[df["customer_visits"] < 0]
df = df[df["customer_visits"] >= 0]
```

```python
df["category"] = df["category"].str.strip().str.title()
```

```python
df["weather_condition"] = df["weather_condition"].str.title()
```

```python
Q1 = df["price"].quantile(0.25)
Q3 = df["price"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR
df = df[(df["price"] >= lower) &
(df["price"] <= upper)]
```

```python
df.to_csv('clean_dynamic_pricing_dataset.csv', index=False)
```
