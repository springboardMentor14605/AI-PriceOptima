```python
import pandas as pd

df = pd.read_csv("clean_dynamic_pricing_dataset.csv")

# Check shape and first 5 rows
print(df.shape)
print(df.head())
```

```python
# Calculate profit margin per product
df["profit_margin"] = df["price"] - df["cost"]

print(df[["price", "cost", "profit_margin"]].head())
```

```python
# Calculate profit margin as a percentage of selling price
df["margin_percent"] = (df["price"] - df["cost"]) / df["price"]

print(df[["price", "cost", "profit_margin", "margin_percent"]].head())
```

```python
# Calculate the ratio of actual sales to forecasted demand
df["demand_ratio"] = df["units_sold"] / (df["demand_forecast"] + 1)

print(df[["units_sold", "demand_forecast", "demand_ratio"]].head())
```

```python
# Calculate inventory pressure
df["inventory_pressure"] = df["units_sold"] / (df["inventory_level"] + 1)

print(df[["units_sold", "inventory_level", "inventory_pressure"]].head())
```

```python
# Calculate the difference between our price and competitor price
df["competitor_gap"] = df["price"] - df["competitor_pricing"]

print(df[["price", "competitor_pricing", "competitor_gap"]].head())
```

```python
# Calculate actual selling price after discount
df["discounted_price"] = df["price"] * (1 - df["discount"])

print(df[["price", "discount", "discounted_price"]].head())
```

```python
# Calculate revenue per product
df["revenue"] = df["units_sold"] * df["discounted_price"]

print(df[["units_sold", "discounted_price", "revenue"]].head())
```

```python
# Calculate price-demand ratio
df["price_demand_ratio"] = df["price"] / (df["units_sold"] + 1)

print(df[["price", "units_sold", "price_demand_ratio"]].head())
```

```python
# Calculate remaining stock per product
df["stock_remaining"] = df["inventory_level"] - df["units_sold"]

# Check first 5 rows
print(df[["inventory_level", "units_sold", "stock_remaining"]].head())
```

```python
# Calculate traffic intensity
df["traffic_intensity"] = df["visitors"] / (df["inventory_level"] + 1)
print(df[["visitors", "inventory_level", "traffic_intensity"]].head())
```

```python
# Ensure 'date' column is datetime type
df["date"] = pd.to_datetime(df["date"])

# Extract day of the week (0 = Monday, 6 = Sunday)
df["day_of_week"] = df["date"].dt.dayofweek

# Extract month (1 = January, 12 = December)
df["month"] = df["date"].dt.month

# Flag weekends (Saturday=5, Sunday=6)
df["is_weekend"] = df["day_of_week"].isin([5, 6]).astype(int)

print(df[["date", "day_of_week", "month", "is_weekend"]].head())
```

```python
df.to_csv("feature_engineered_dataset.csv", index=False)
```
